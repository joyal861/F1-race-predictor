```python
# Cell 1: Import Libraries
import fastf1
import pandas as pd
import os

print("All libraries imported successfully!")
print(f"FastF1 version: {fastf1.__version__}")
```

```python
# Cell 2: Set up cache
cache_path = "../data/raw/cache"
os.makedirs(cache_path, exist_ok=True)

fastf1.Cache.enable_cache(cache_path)
print(f"Cache enabled at: {cache_path}")
```

```python
# Cell 3: Load 2023 Australian Grand Prix race session
session = fastf1.get_session(2023, 'Australia', 'R')
session.load()

print("Session loaded!")
print(f"Event: {session.event['EventName']}")
print(f"Date: {session.event['EventDate']}")
print(f"Circuit: {session.event['Location']}")
```

```python
# Cell 4: Explore race results
results = session.results

results_clean = results[[
    'DriverNumber', 'Abbreviation', 'FullName',
    'TeamName', 'GridPosition', 'Position',
    'Points', 'Status'
]]

print(f"Number of drivers: {len(results_clean)}")
print("\nRace Results:")
results_clean
```

```python
# Cell 5: Save to CSV
output_path = "../data/raw/2023_australia_race_results.csv"
results_clean.to_csv(output_path, index=False)
print(f"Data saved to: {output_path}")
print(f"Shape: {results_clean.shape[0]} rows, {results_clean.shape[1]} columns")
```

```python
# Cell 6: Collect multiple races across seasons
# We'll collect a full season worth of races

races_to_collect = [
    (2023, 'Bahrain'),
    (2023, 'Saudi Arabia'),
    (2023, 'Australia'),
    (2023, 'Azerbaijan'),
    (2023, 'Miami'),
    (2023, 'Monaco'),
    (2023, 'Spain'),
    (2023, 'Canada'),
    (2023, 'Britain'),
    (2023, 'Hungary'),
]

all_results = []

for year, race in races_to_collect:
    try:
        print(f"Loading {year} {race}...")
        s = fastf1.get_session(year, race, 'R')
        s.load(telemetry=False, weather=False, messages=False)
        
        r = s.results[['DriverNumber', 'Abbreviation', 'FullName',
                        'TeamName', 'GridPosition', 'Position',
                        'Points', 'Status']].copy()
        r['Year'] = year
        r['Race'] = race
        all_results.append(r)
        print(f"  ✓ {len(r)} drivers loaded")
        
    except Exception as e:
        print(f"  ✗ Failed: {e}")

# Combine all races into one dataframe
df = pd.concat(all_results, ignore_index=True)
print(f"\nTotal records collected: {len(df)}")
df.head(10)
```

```python
# Cell 7: Save combined dataset
output_path = "../data/raw/2023_races_combined.csv"
df.to_csv(output_path, index=False)

print(f"Dataset saved!")
print(f"Total rows: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")
print(f"\nRaces collected: {df['Race'].unique()}")
print(f"\nColumn names: {list(df.columns)}")
```

```python
# Cell 8: Understand our data better
print("=== DATA OVERVIEW ===")
print(f"\nShape: {df.shape}")
print(f"\nData types:\n{df.dtypes}")
print(f"\nMissing values:\n{df.isnull().sum()}")
print(f"\nSample rows:")
df.sample(5)
```
